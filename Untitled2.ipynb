```python
import google.generativeai as genai
from google.colab import userdata

genai.configure(api_key=userdata.get('GEMINI_KEYS'))
```

```python
model = genai.GenerativeModel(
    model_name='gemini-2.5-flash-lite',
    system_instruction="you are a helpul assistant."
)

chat = model.start_chat(history=[]) # fresh chat
```

```python
r1=model.generate_content('my name is rakesh')
print ("bot",r1.text)
```

```python
reply = chat.send_message("Hey, what is my name?")
print("Bot with history", reply.text)
```

```python
model = genai.GenerativeModel(
    model_name='gemini-2.5-flash-lite',
    system_instruction="you are friendly, helpful assitant. keep replies short"
    "(under 80 words) unless the user asks for more details"
)

chat = model.start_chat(history=[])

print("Bot is ready!, Type quit, bye , exit ")

while True:
  user_input = input("You: ")
  if user_input.lower() in ["quit", "bye", "exit"]:
    print("Bot: Goodbye!")
    break

  reply = chat.send_message(user_input)
  print(f"Bot: {reply.text}\n")
```

```python
print("========")
for i, msg in enumerate(chat.history):
  role = "User" if msg.role == "user" else " BOT"
  print(f"{i+1}. {role}: {msg.parts[0].text}")
```
