```python
from google.ai.generativelanguage_v1beta.types.generative_service import GenerationConfig
import google.generativeai as genai
from google.colab import userdata
genai.configure(api_key=userdata.get('GEMINI_KEYS'))
def generate_blog(topic, genre, length_words=200):
    system_prompt=(
        " you are a creative blog writer"
        "and posts news on everything.you write in clean and concise and easy to read in english."
    )
    user_prompt=f"""
    write a new blog.
    topic:{topic}
    Genre:{genre}
    Length: approximatelyy {length_words} words
    """
    model = genai.GenerativeModel(
        model_name='models/gemini-flash-latest',
        system_instruction=system_prompt
    )
    response = model.generate_content(
        user_prompt,
        generation_config={
            "temperature":1.0,
            "max_output_tokens":2000,
           # "top_k":40,
           # "top_p":0.9
        }
    )
    return response.text
```

```python
blog = generate_blog(
    topic='about industrial raise in india',
    genre="indian industries",
    length_words=2000

)
print(blog)
```

```python
import google.generativeai as genai
from google.ai.generativelanguage_v1beta.types.generative_service import GenerationConfig
from google.colab import userdata

# Configure the SDK with your API key
genai.configure(api_key=userdata.get('GEMINI_KEYS'))

def generate_email(recipient, purpose, tone, key_points, length_words=150):
    """
    Generates a professional or casual email using Gemini.
    """
    system_prompt = (
        "You are an expert administrative assistant and professional communicator. "
        "You write clean, concise, polite, and effective emails in flawless English."
    )

    user_prompt = f"""
    Write an email based on the following details:
    Recipient: {recipient}
    Purpose/Subject: {purpose}
    Tone: {tone}
    Key Points to Include: {key_points}
    Length: Approximately {length_words} words.

    Please include a clear Subject Line, appropriate greeting, body, and sign-off.
    """


    model = genai.GenerativeModel(
        model_name='models/gemini-flash-latest',
        system_instruction=system_prompt
    )

    # Generate the email content
    response = model.generate_content(
        user_prompt,
        generation_config={
            "temperature": 0.7,
            "max_output_tokens": 1500,
        }
    )

    return response.text
```

```python

email = generate_email(
    recipient="The Management Team / Shareholders",
    purpose="An overview of the industrial rise in India",
    tone="Informative and professional",
    key_points="Rapid growth in manufacturing, government initiatives like Make in India, rise of tech-driven industries, and future economic outlook.",
    length_words=300 # Adjusted to a realistic email length
)

print(email)
```
